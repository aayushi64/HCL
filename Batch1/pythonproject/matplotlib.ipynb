```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([10,20,30,40])
y=np.array([100,200,300,400])
plt.plot(x,y,marker="*")
plt.show()
```

```python
products=['mobile','pendrive','laptop','watch']
sales=[3000,1500,4000,2500]
plt.bar(products,sales,width=0.5,color='aqua')
plt.show()
```

```python
products=['mobile','pendrive','laptop','watch']
sales=[3000,1500,4000,2500]
plt.barh(products,sales,height=0.5,color='aqua')
plt.xlabel('sales')
plt.ylabel('product')
plt.title('product wise sales')
plt.show()
```

```python
products=['mobile','pendrive','laptop','watch']
prices=[3000,1500,4000,2500]
plt.pie(prices,labels=products)

plt.legend()

plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
data=pd.read_excel('Superstore.xlsx')
data.head()
```

```python
data
```

```python
count= data['Region'].unique()
count
```

```python
count= data['Category'].nunique()
count
```

```python
count= data['Sub-Category'].nunique()
count
```

```python
category_sales= data.groupby('Category')['Sales'].sum()
plt.figure(figsize=(4,3))
category_sales.plot(kind='bar', color='red')
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Horizontal Bar Chart: Region vs Sales
plt.figure(figsize=(8, 5))
sns.barplot(data=data, x='Sales', y='Region', errorbar=None)
plt.title('Sales by Region')
plt.show()
```

```python
# 2. Scatter Plot: Sub-Category vs Sales and Profit
plt.figure(figsize=(10, 5))
sns.scatterplot(data=data, x='Sales', y='Profit', hue='Sub-Category')
plt.title('Sales vs Profit by Sub-Category')
plt.show()
```

```python
# 3. Pie Chart: Segment vs Sales
segment_sales = data.groupby('Segment')['Sales'].sum()

plt.figure(figsize=(7, 7))
plt.pie(segment_sales, labels=segment_sales.index, autopct='%1.1f%%')
plt.title('Sales Distribution by Segment')
plt.show()
```
