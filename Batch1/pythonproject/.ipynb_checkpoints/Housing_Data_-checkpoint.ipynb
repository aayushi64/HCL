### Summing up all null values column-wise

```python
# data.isnull()  # It will return True where value is Null
# data.isnull().sum()   # It will return the sum of all Null values column-wise
data.isnull().sum().sort_values(ascending = False)   # Sorting Values in Descending Order
```

```python
# data.isnull()  # It will return True where value is Null
# data.isnull().sum(axis=1)   # It will return the sum of all Null values column-wise
data.isnull().sum(axis=1).sort_values(ascending = False)   # Sorting Values in Descending Order
```

### Showing Columns & Rows having all null values

```python
data.isnull().all()  # Show Columns having all null values
```

```python
data.isnull().all(axis=1)  # Show Rows having all null values
```

### Showing Columns and Rows having atleast one null value

```python
data.isnull().any(axis=0) # Retuns all the colunms True having atleast one missing value
```

```python
data.isnull().any(axis=1) # Retuns all the rows True having atleast one missing value
```

### Summing up the missing values (column-wise) - in %

```python
data.isnull().sum().sort_values(ascending = False)/len(data) * 100
round(data.isnull().sum().sort_values(ascending = False)/len(data) * 100, 2)
```

### Removing the Top 2 columns where the max null values %

```python
data = data.drop(['BuildingArea', 'YearBuilt'], axis=1)   # Drop method is used to Delete the columns from Data
```

```python
round(data.isnull().sum().sort_values(ascending = False)/len(data) * 100, 2)
```

### Check the rows having more than 5 missing values

```python
data.isnull().sum(axis=1)>5
data[data.isnull().sum(axis=1)>5]
len(data[data.isnull().sum(axis=1)>5])/len(data) * 100
```

### Retaining the rows having less than 5 missing values

```python
data[data.isnull().sum(axis=1)<5]
len(data[data.isnull().sum(axis=1)<5])/len(data) * 100
```

```python
data = data[data.isnull().sum(axis=1)<5]
data
```

```python
round(data.isnull().sum().sort_values(ascending = False)/len(data) * 100, 2)
```

```python
data['Price'].notnull()
data[data['Price'].notnull()]
```

```python
data = data[data['Price'].notnull()]
data
```

```python
round(data.isnull().sum().sort_values(ascending = False)/len(data) * 100, 2)
```
