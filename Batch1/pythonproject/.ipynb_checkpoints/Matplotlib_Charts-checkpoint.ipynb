```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([18,20,30,40])
y = np.array([100,200,300,400])
plt.plot(x,y,marker= "*")
plt.show()
```

```python
products = ['Mobile', 'Pendrive','Laptop', 'Watch']
sales = [3000,2000,6000,4500]
plt.bar(products,sales,width=0.5,color = 'red')
plt.show()
```

```python
products = ["Mobile", "Pendrive", "Laptop","Watch"]
sales = [200,300,450,433]
plt.barh(products,sales,height = 0.5,color = "navy")
plt.xlabel('Sales')
plt.ylabel('Products')
plt.title("sales via customer")
plt.show()
```

```python
products = ["Mobile", "Pendrive", "Laptop", "Watch"]
sales = [200, 300, 450, 433]

plt.pie(sales, labels=products)
plt.legend()
plt.show()
```
