```python
df=pd.read_excel("Superstore.xlsx")
df["Order Date"]=pd.to_datetime(df["Order Date"])
df["Ship Date"]=pd.to_datetime(df["Ship Date"])
df["Year"] =df["Order Date"].dt.year
df["Month"] =df["Order Date"].dt.month
df["MonthName"] =df["Order Date"].dt.strftime("%b")
month_order=["Jan","Feb"," Mar","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
df.head()
```

```python
top10= (df.groupby("Sub-Category")["Sales"].sum().nlargest(10).sort_values())
plt.figure(figsize=(9,5))
plt.barh(top10.index,top10.values/1000, color="steelblue")
plt.title("top 10 sub-categoirees by total sales")
plt.xlabel("total sale (in thousand of $)")
plt.ylabel("sub-category")
plt.tight_layout()
plt.show()
```
