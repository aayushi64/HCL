```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
np.random.seed(42)
n = 40
df = pd.DataFrame({
    "Student_ID": range(1, n + 1),
    "Branch": np.random.choice(["CSE", "ECE", "ME", "CE"], n),
    "Study_Hours": np.round(np.random.uniform(1, 8, n), 1),
    "Attendance": np.random.randint(55, 100, n),
})
df["Marks"] = np.clip(
    np.round(30 + 5 * df["Study_Hours"] + 0.2 * df["Attendance"] + np.random.normal(0, 6, n)),
    0, 100
).astype(int)

df.head()
```

# Task 1: Create a function to assign grades based on marks

- > Write a function get_grade(marks) that returns "A" for 75 and above, "B" for 60 to 74, "C" for 40 to 59, and "F" below 40.
- > Add a Grade column to df using apply().
- > Show how many students got each grade using value_counts().


```python
def get_grade(marks):
    if marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "F"

# Apply the grading function to create a new Grade column
df["Grade"] = df["Marks"].apply(get_grade)

# Count the number of students in each grade category
grade_counts = df["Grade"].value_counts()

print("Grade Counts:")
print(grade_counts)

# Total number of students
print("\nTotal Students:", grade_counts.sum())

# Display first 5 rows
df.head()
```

# Task 2: Create a list of marks for students with attendance below 75

- >	Using a list comprehension, create a list of marks of students whose attendance is below 75.
- >	Using filter() and a lambda, get all marks above 70.
- >	Using map() and a lambda, add 5 grace marks to every student's marks (the maximum must stay 100).


```python
low_attendance_marks = [
    marks
    for marks, attendance in zip(df["Marks"], df["Attendance"])
    if attendance < 75
]

print("Marks of students with attendance below 75:")
print(low_attendance_marks)


# Use filter() and lambda to get all marks above 70
marks_above_70 = list(filter(lambda x: x > 70, df["Marks"]))

print("\nMarks above 70:")
print(marks_above_70)


# Use map() and lambda to add 5 grace marks (maximum 100)
grace_marks = list(map(lambda x: min(x + 5, 100), df["Marks"]))

print("\nMarks after adding 5 grace marks:")
print(grace_marks)
```

# Task 3: Flexible functions

- >	Write a function summary(*values, round_to=2) that accepts any number of values and returns a dictionary with the count, minimum, maximum and average.
- >	Test it with summary(10, 20, 30) and with summary(*df["Marks"]).


```python
def summary(*values, round_to=2):
    return {
        "count": len(values),
        "minimum": min(values),
        "maximum": max(values),
        "average": round(sum(values) / len(values), round_to)
    }

# Test the function with sample values
print(summary(10, 20, 30))

# Test the function with all marks from the DataFrame
print(summary(*df["Marks"]))
```

# Task 5: Find average Marks and Study_Hours for each Branch


- >Find the average Marks and average Study_Hours for each Branch using groupby().
- >Which branch has the highest average marks?
- >Display the top 5 students by marks using sort_values().


```python
branch_summary = df.groupby("Branch")[["Marks", "Study_Hours"]].mean()

print("Average Marks and Study Hours by Branch:")
print(branch_summary)


# Find the branch with the highest average marks
top_branch = branch_summary["Marks"].idxmax()

print("\nBranch with the highest average marks:")
print(top_branch)


# Display the top 5 students by marks
top_5_students = df.sort_values(by="Marks", ascending=False).head(5)

print("\nTop 5 Students by Marks:")
print(top_5_students)
```

# Task 6: Convert the Marks column to a NumPy array


- >Convert the Marks column to a NumPy array.
- >Calculate the z-score of every student using the formula (marks − mean) ÷ standard deviation. Do not use a loop.
- >How many students have a z-score greater than 1?


```python
marks = df["Marks"].to_numpy()

# Calculate z-scores without using a loop
mean_marks = np.mean(marks)
std_marks = np.std(marks)

z_scores = (marks - mean_marks) / std_marks

print("Z-scores:")
print(z_scores)


# Count students with z-score greater than 1
count = np.sum(z_scores > 1)

print("\nNumber of students with z-score > 1:")
print(count)
```

# Task 7: Calculate descriptive statistics for Marks


- >Calculate the mean, median, mode, variance, standard deviation, minimum and maximum of Marks.
- >Find Q1, Q3 and the IQR. Use the rule lower limit = Q1 − 1.5 × IQR and upper limit = Q3 + 1.5 × IQR to check for outliers.
- >In 2 to 3 lines, explain what the mean, median and standard deviation tell you about the class.


```python
marks = df["Marks"]

print("Mean:", marks.mean())
print("Median:", marks.median())
print("Mode:", marks.mode().tolist())
print("Variance:", marks.var())
print("Standard Deviation:", marks.std())
print("Minimum:", marks.min())
print("Maximum:", marks.max())


# Find Q1, Q3, and IQR
Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)
IQR = Q3 - Q1

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)


# Calculate lower and upper limits for outliers
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)


# Find outliers
outliers = marks[(marks < lower_limit) | (marks > upper_limit)]

print("\nOutliers:")
print(outliers)

print("\nNumber of Outliers:", len(outliers))
```

- >Mean shows the average marks of all students and gives an overall idea of class performance.
- >Median represents the middle score and helps understand the typical performance without being affected much by extreme values.
- >Standard Deviation indicates how spread out the marks are from the average; a larger value means greater variation among students' scores.

# Task 9: Calculate a 95% confidence interval for the average marks


- >Calculate a 95% confidence interval for the average marks using stats.t.interval().
- >Write one sentence explaining what this interval means.


```python
marks = df["Marks"]

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(marks) - 1,
    loc=marks.mean(),
    scale=stats.sem(marks)
)

print("95% Confidence Interval:")
print(confidence_interval)
```

We are 95% confident that the true average marks of all students lie within this confidence interval

# Task 10: Perform a one-sample t-test


- >  H0 (Null Hypothesis): The average marks of students is 60.
- > H1 (Alternative Hypothesis): The average marks of students is not 60.

```python
t_statistic, p_value = stats.ttest_1samp(df["Marks"], 60)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

# Decision based on significance level 0.05
alpha = 0.05

if p_value < alpha:
    print("\nReject H0")
    print("There is sufficient evidence that the average marks are different from 60.")
else:
    print("\nFail to Reject H0")
    print("There is not enough evidence to say that the average marks are different from 60.")
```

- > **Hypotheses**

H₀ (Null Hypothesis): μ = 60
 (The average marks of students is 60.)
H₁ (Alternative Hypothesis): μ ≠ 60
 (The average marks of students is not 60.)

- > **Interpretation**

If p-value < 0.05, reject H₀. This means the data suggests that the average marks are significantly different from 60.
If p-value ≥ 0.05, fail to reject H₀. This means there is not enough evidence to conclude that the average marks differ from 60.

- > Since the p-value is less than 0.05, we reject the null hypothesis and conclude that the average marks of students are significantly different from 60.

# Task 11: Compare the marks of CSE and ECE students using an independent t-test

- >  H0: The average marks of CSE and ECE students are equal.
- > H1: The average marks of CSE and ECE students are different

```python
cse_marks = df[df["Branch"] == "CSE"]["Marks"]
ece_marks = df[df["Branch"] == "ECE"]["Marks"]

t_statistic, p_value = stats.ttest_ind(cse_marks, ece_marks)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

# Decision at 0.05 significance level
alpha = 0.05

if p_value < alpha:
    print("\nReject H0")
    print("There is a significant difference between the average marks of CSE and ECE students.")
else:
    print("\nFail to Reject H0")
    print("There is no significant difference between the average marks of CSE and ECE students.")
```

- >If p-value < 0.05 → Reject H₀. There is a significant difference between the average marks of CSE and ECE students.
- >If p-value ≥ 0.05 → Fail to Reject H₀. There is no significant difference between the average marks of CSE and ECE students.
