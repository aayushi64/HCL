```python
# adding two list using loops
num1 = [1,2,3,4,5,6]
num2 = [6,5,4,3,2,1]
num3 = [num1[i] + num2[i] for i in range(len(num1))]
print(num3)
#For every index i in the list, add num1[i] and num2[i] and store the result
```

```python
#find the sum of all digits in a number 
n = int(input("Enter a number: "))

total = 0

for digit in str(n):
    total += int(digit)

print("Sum of digits:", total)
```

```python
num = int(input("Enter a number: "))
sum = 0
while num > 0:
    digit = num % 10
    sum += digit
    num //= 10

print("Sum of digits:", sum)
```
